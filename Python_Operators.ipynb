```python
# 3. Find whether a number is even or odd using the modulus operator.
number=int(input("Enter a number: "))
if number%2==0:
    print("Even")
else:
    print("Odd")
```

```python
# 4. Reverse a three-digit number using arithmetic operators.
number=int(input("Enter a three digit number: "))
Reverse_number=0
while number>0:
    remainder=number%10
    Reverse_number=(Reverse_number*10)+remainder
    number=number//10
print("Reverse number",Reverse_number)
```

```python
# 5. Find the last digit and first digit of a given number.
number = int(input("Enter a number: "))
last_digit = number % 10
temp = number
while temp >= 10:
    temp = temp // 10
first_digit = temp

print("First digit:", first_digit)
print("Last digit:", last_digit)
```

```python
# 6. Demonstrate all assignment operators (`=`, `+=`, `-=`, `*=`, `/=`, `//=`, `%=` and `**=`) on a variable.
num = 10
print("Initial value:", num)

num += 5
print("After += 5:", num)

num -= 3
print("After -= 3:", num)

num *= 2
print("After *= 2:", num)

num /= 4
print("After /= 4:", num)

num //= 2
print("After //= 2:", num)

num %= 3
print("After %= 3:", num)

num **= 2
print("After **= 2:", num)
```

```python
# 17. Determine whether a year is a leap year using logical conditions.
year=int(input("Enter a year"))
if year%4==0 and year%100!=0 or year%400==0:
  print("Leap year")
else:
  print("Not a leap year")
```

```python
# 27. Count how many vowels in a string belong to a predefined vowel list using `in`.
sentence=str(input("Enter a sentence"))
vowels=['a','e','i','o','u']
count=0
for char in sentence:
  if char in vowels:
    count+=1
print("count",count)
```

```python
# 28. Build a simple calculator using arithmetic and comparison operators.
num1 = float(input("Enter first number: "))
operator = input("Enter operator (+, -, *, /, %): ")
num2 = float(input("Enter second number: "))

if operator == "+":
    print("Result:", num1 + num2)
elif operator == "-":
    print("Result:", num1 - num2)
elif operator == "*":
    print("Result:", num1 * num2)
elif operator == "/":
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Division by zero is not allowed.")
elif operator == "%":
    if num2 != 0:
        print("Result:", num1 % num2)
    else:
        print("Modulo by zero is not allowed.")
else:
    print("Invalid operator.")
```

```python
# 29. Create a grading system using comparison and logical operators.
marks = float(input("Enter your marks: "))

if marks >= 90 and marks <= 100:
    print("Grade: A")

elif marks >= 80 and marks < 90:
    print("Grade: B")

elif marks >= 70 and marks < 80:
    print("Grade: C")

elif marks >= 60 and marks < 70:
    print("Grade: D")

elif marks >= 0 and marks < 60:
    print("Grade: F")

else:
    print("Invalid Marks")
```

```python
# 30. Build a menu-driven calculator that uses arithmetic, comparison, logical, assignment, membership, identity, and bitwise operators wherever applicable.
while True:
    print("1. Arithmetic Operations")
    print("2. Comparison Operations")
    print("3. Logical Operations")
    print("4. Assignment Operators")
    print("5. Membership Operators")
    print("6. Identity Operators")
    print("7. Bitwise Operators")
    print("8. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        num1 = int(input("Enter first number: "))
        num2 = int(input("Enter second number: "))

        print("Addition:", num1 + num2)
        print("Subtraction:", num1 - num2)
        print("Multiplication:", num1 * num2)

        if num2 != 0:
            print("Division:", num1 / num2)
            print("Modulus:", num1 % num2)
        else:
            print("Division by zero is not allowed.")

    elif choice == 2:
        num1 = int(input("Enter first number: "))
        num2 = int(input("Enter second number: "))

        print("num1 == num2 :", num1 == num2)
        print("num1 != num2 :", num1 != num2)
        print("num1 > num2  :", num1 > num2)
        print("num1 < num2  :", num1 < num2)
        print("num1 >= num2 :", num1 >= num2)
        print("num1 <= num2 :", num1 <= num2)

    elif choice == 3:
        age = int(input("Enter age: "))
        citizen = input("Are you a citizen? (yes/no): ")

        if age >= 18 and citizen == "yes":
            print("Eligible to vote")
        else:
            print("Not eligible to vote")

    elif choice == 4:
        balance = 1000
        print("Initial Balance:", balance)

        balance += 500
        print("After Deposit:", balance)

        balance -= 200
        print("After Withdrawal:", balance)

    elif choice == 5:
        vowels = ['a', 'e', 'i', 'o', 'u']
        ch = input("Enter a character: ").lower()

        if ch in vowels:
            print("It is a vowel.")
        else:
            print("It is not a vowel.")

    elif choice == 6:
        list1 = [10, 20, 30]
        list2 = list1

        print("list1 is list2 :", list1 is list2)
        print("list1 is not list2 :", list1 is not list2)

    elif choice == 7:
        num1 = int(input("Enter first number: "))
        num2 = int(input("Enter second number: "))

        print("AND :", num1 & num2)
        print("OR  :", num1 | num2)
        print("XOR :", num1 ^ num2)
        print("Left Shift :", num1 << 2)
        print("Right Shift:", num1 >> 2)

    elif choice == 8:
        print("Thank You!")
        break

    else:
        print("Invalid Choice")
```
