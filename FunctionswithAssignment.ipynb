Assignment 32: Function Returning a Dictionary
Write a function named analyze_text().

Requirements
Accept a string.
Return a dictionary containing:
Total characters
Total words
Total vowels
Total digits
Total spaces
Display every result clearly.

```python
def analyze_text(text):
  vowels="aeiouAEIOU"
  characters_count=len(text)
  words_count=len(text.split())
  vowels_count=0
  digits_count=0
  space_count=0
  for char in text:
    if char in vowels:
      vowels_count+=1
    if char.isdigit():
      digits_count+=1
    if char == " ":
      space_count+=1
  return {
      "total_characters":characters_count,
      "total_words":words_count,
      "total_vowels":vowels_count,
      "total_digits":digits_count,
      "total_spaces":space_count
  }
text=str(input("enter your sentence:"))
result=analyze_text(text)
print(f"Total characters: {result['total_characters']}")
print(f"Total words: {result['total_words']}")
print(f"Total vowels: {result['total_vowels']}")
print(f"Total digits: {result['total_digits']}")
print(f"Total spaces: {result['total_spaces']}")
```

```python
def is_prime(num):
  if num<=1:
    return False
  else:
    for i in range(2,num):
      if num%i==0:
        return False
    return True
num=int(input("enter number:"))
if is_prime(num):
  print(f"{num} is a prime number")
else:
  print(f"{num} is not a prime number")
```

Assignment 35: Fibonacci Function

Return the complete Fibonacci series as a list.
Handle zero and negative term counts.
Do not use recursion in this assignment.
Example Input
Enter number of terms: 7
Expected Output
[0, 1, 1, 2, 3, 5, 8]

```python
def generate_fibonacci(num_terms):
  if num_terms<=0:
    return []
  elif num_terms==1:
    return [0]
  else:
    fibonacci_list=[0,1]
    for i in range(2,num_terms):
      fibonacci_list.append(fibonacci_list[-1]+fibonacci_list[-2])
    return fibonacci_list
num_terms=int(input("enter number of terms:"))
fibonacci_list=generate_fibonacci(num_terms)
print(f"{fibonacci_list}")
```

Assignment 92: Closure-Based Bank Balance
Write a function named create_bank_account().

Requirements
Store the balance in the outer function.
Define inner functions for:
Deposit
Withdrawal
Balance enquiry
Use nonlocal to update the balance.
Reject withdrawals greater than the available balance.
Return all three inner functions.
Hint
Return multiple functions:

return deposit, withdraw, check_balance

```python
def create_bank_account(initial_balance):
    balance = initial_balance

    def deposit(amount):
        nonlocal balance
        balance += amount
        return balance

    def withdraw(amount):
        nonlocal balance
        if amount <= balance:
            balance -= amount
            return balance
        else:
            return "Insufficient Balance"

    def check_balance():
        return balance

    return deposit, withdraw, check_balance

deposit, withdraw, check_balance = create_bank_account(5000)

print("Current Balance:", check_balance())

print("After Deposit:", deposit(2000))

print("After Withdrawal:", withdraw(1500))

print("Current Balance:", check_balance())

print("Attempt Large Withdrawal:", withdraw(10000))

print("Final Balance:", check_balance())
```

It should print another message after execution.

Preserve the original function's return value.
Expected Pattern
Starting function...
Result: 30
Function completed.

```python
def execution_logger(function):

    def wrapper(a, b):
        print("Starting function")

        result = function(a, b)

        print("Function completed")

        return result

    return wrapper


@execution_logger
def add(num1, num2):
    return num1 + num2


result = add(10, 20)

print("Result:", result)
```
