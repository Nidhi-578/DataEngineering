```python
#90.while loop -Coin change-Greedy count
amount = int(input("Enter the amount: "))
coins = [100, 50, 20, 10, 5, 2, 1]
coin_count = {}
i = 0
while i < len(coins):
    coin = coins[i]
    count = amount // coin
    if count > 0:
        coin_count[coin] = count
    amount = amount % coin
    i += 1
print(coin_count)

```

```python
#91. Binary search with while
numbers = [10, 20, 30, 40, 50, 60, 70, 80, 90]
target = int(input("Enter the number to search: "))
left = 0
right = len(numbers) - 1
while left <= right:
    mid = (left + right) // 2
    if numbers[mid] == target:
        print("Element found at index", mid)
        break
    elif numbers[mid] < target:
        left = mid + 1
    else:
        right = mid - 1
else:
    print("Element not found")
```

```python
# 96. Queue Simulation
queue = []

while True:
    print("1. Enqueue")
    print("2. Dequeue")
    print("3. Display")
    print("4. Exit")
    choice = input("Enter your choice: ")
    if choice == "1":
        item = input("Enter element: ")
        queue.append(item)
        print(item, "added to queue")
    elif choice == "2":
        if len(queue) == 0:
            print("Queue is empty")
        else:
            removed = queue.pop(0)
            print(removed, "removed from queue")
    elif choice == "3":
        print("Queue:", queue)
    elif choice == "4":
        print("program ended")
        break
    else:
        print("invalid choice")
```
