```python
# 12. Find the quotient and remainder using `divmod()`.
num = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
if num2 != 0:
    quotient, remainder = divmod(num, num2)
    print("Quotient:", quotient)
    print("Remainder:", remainder)
else:
    print("Division by zero is not allowed.")
```

```python
# 13. Calculate the sine, cosine, and tangent of 45 degrees.
import math
angle = float(input("Enter the angle in degrees: "))
radian = math.radians(angle)
sine = math.sin(radian)
cosine = math.cos(radian)
tangent = math.tan(radian)
print("Sine:", sine)
print("Cosine:", cosine)
print("Tangent:", tangent)
```

```python
# 20. Find the distance between two points using `math.dist()`.
import math
point1 = (3, 4)
point2 = (6, 8)
distance = math.dist(point1, point2)
print("Distance between the points:", distance)
```

```python
# 21. Calculate the hypotenuse of a right triangle using `math.hypot()`.
import math
base = float(input("Enter the base length of the triangle: "))
height = float(input("Enter the height length of the triangle: "))
hypotenuse = math.hypot(base, height)
print("Hypotenuse of the triangle:", hypotenuse)
```

```python
# 22. Find the combination (nCr) using `math.comb()`.
import math
n = int(input("Enter the value of n: "))
r = int(input("Enter the value of r: "))
combination = math.comb(n, r)
print("Combination (nCr):", combination)
```

```python
# 24. Check whether a number is finite, infinite, or NaN using `math.isfinite()`, `math.isinf()`, and `math.isnan()`.
num=float(input("Enter a number: "))
print(math.isfinite(num))
print(math.isinf(num))
print(math.isnan(num))
```

```python
# 25. Find the product of all numbers in a list using `math.prod()`.
import math
num=list(map(int,input("enter numbers separated by spaces").split()))
print(math.prod(num))
```

```python
# 26. Find the sum of square roots of all numbers in a list.
num=list(map(int,input("enter numbers separated by spaces").split()))
print(sum(math.sqrt(i) for i in num))
```

```python
# 27. Find the largest integer less than or equal to the square root of a given number.
num=int(input("Enter a number: "))
print(math.floor(math.sqrt(num)))
```

```python
# 28. Calculate compound interest using `math.pow()`.
p = float(input("Enter the principal amount: "))
r = float(input("Enter the annual interest rate: "))
t = float(input("Enter the time period in years: "))
n = int(input("Enter the number of times interest is compounded per year: "))
amount = p * math.pow((1 + (r / (100 * n))), n * t)
compound_interest = amount - p
print("Compound Interest:", compound_interest)
```

```python
# 29. Find the area and circumference of a circle using `math.pi`.
import math
radius = float(input("Enter the radius of the circle: "))
area = math.pi * radius**2
circumference = 2 * math.pi * radius
print("Area of the circle:", area)
print("Circumference of the circle:", circumference)
```

```python
# 30. Create a scientific calculator that supports square root, factorial, logarithm, power, trigonometric functions, GCD, and LCM.
import math
while True:
    print("Scientific Calculator")
    print("1. Square Root")
    print("2. Factorial")
    print("3. Logarithm")
    print("4. Power")
    print("5. Trigonometric Functions")
    print("6. GCD")
    print("7. LCM")
    print("8. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        num = float(input("Enter a number: "))
        print("Square Root:", math.sqrt(num))
    elif choice == 2:
        num = int(input("Enter a number: "))
        print("Factorial:", math.factorial(num))
    elif choice == 3:
        num = float(input("Enter a number: "))
        print("Logarithm:", math.log(num))
    elif choice == 4:
        base = float(input("Enter a number: "))
        exponent = float(input("Enter the exponent: "))
        print("Power:", math.pow(base, exponent))
    elif choice == 5:
        angle = float(input("Enter the angle in degrees: "))
        radian = math.radians(angle)
        print("Sine:", math.sin(radian))
        print("Cosine:", math.cos(radian))
        print("Tangent:", math.tan(radian))
    elif choice == 6:
        num1 = int(input("Enter the first number: "))
        num2 = int(input("Enter the second number: "))
        print("GCD:", math.gcd(num1, num2))
    elif choice == 7:
        num1 = int(input("Enter the first number: "))
        num2 = int(input("Enter the second number: "))
        print("LCM:", math.lcm(num1, num2))
    elif choice == 8:
        print("Exiting the calculator.")
        break
    else:
        print("Invalid choice! Please enter a number between 1 and 8.")



```
